```python
from sentence_transformers import SentenceTransformer

# Load the BGE base model
model = SentenceTransformer('BAAI/bge-base-en-v1.5')

# Example sentences
sentences = [
    "This is an example sentence.",
    "Another sentence for embedding.",
    "And a third one."
]

# Generate embeddings
embeddings = model.encode(sentences, normalize_embeddings=True)

# Print the embeddings
for sentence, embedding in zip(sentences, embeddings):
    print(f"Sentence: {sentence}")
    print(f"Embedding shape: {embedding.shape}")
    print("Embedding:", embedding)
```

```python
# in this example, the word2vec is trained on 3 sample sentences. Once it is trained, neural embedding/vector for a word can be fetched. The same model can also be used as model.load to get the neural emebedding for that word and the neural embedding/vector will be same as before

# !pip install --upgrade numpy
# !pip install --upgrade gensim

from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk

# Download the necessary NLTK data package
nltk.download('punkt_tab')  # Download Punkt sentence tokenizer models
nltk.download('punkt')  # Download the Punkt tokenizer models

# Sample sentences
sentences = [
    "This is the first apple",
    "This is the second sentence",
    "Yet another sentence"
]

# Tokenize the sentences
tokenized_sentences = [word_tokenize(sentence.lower()) for sentence in sentences]

# Train the Word2Vec model
model = Word2Vec(sentences=tokenized_sentences, vector_size=100, window=5, min_count=1, sg=0)

# Get the vector for a word
vector = model.wv['sentence']
print(f"Vector for 'sentence': {vector}")

# Find similar words
# similar_words = model.wv.most_similar('second', topn=3)
# print(f"Words similar to 'sentence': {similar_words}")

# # Save the model
# model.save("word2vec.model")

# # Load the model
loaded_model = Word2Vec.load("word2vec.model")

# Use the loaded model
vector_loaded = loaded_model.wv['sentence']
print(f"Vector for 'sentence' from loaded model: {vector_loaded}")
```
